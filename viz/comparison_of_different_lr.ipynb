```python
import matplotlib.pyplot as plt
```

```python
width = [95.08, 97.01, 97.72, 96.3]
y = ["constant", "adaptive", "invscaling", "optimal"]
colors = ["tab:olive" ,"tab:orange", "tab:cyan", "tab:pink"]
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_title("Comparison of Different Learning Rate Settings")
ax.set_ylabel("Cross-validated Accuracy on Breast Cancer Dataset")
rects = ax.barh(y, width, label=width, color=colors)
ax.invert_yaxis()
for rect in rects:
    width = rect.get_width()
    ax.text(1.05* rect.get_width(), rect.get_y()+0.5*rect.get_height(),"%.2f"%width, ha='center', va='center')
ax.set_xmargin(0.15)
fig.tight_layout()
plt.savefig("comparison_of_different_lr.png", dpi=300)
plt.show()
```
